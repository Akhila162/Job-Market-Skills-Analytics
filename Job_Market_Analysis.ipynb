```python
skills = (
    df['tagsAndSkills']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
)

top_skills = skills.value_counts().head(20)
top_skills
```

```python
top_skills.plot(
    kind='bar',
    figsize=(12, 6),
    title='Top 20 Most Demanded Skills'
)
```

```python
cse_skills = [
    'Python', 'Java', 'SQL', 'JavaScript', 'HTML', 'CSS',
    'React', 'Node.js', 'AWS', 'Azure', 'Cloud',
    'Docker', 'Kubernetes', 'Git', 'GitHub',
    'Machine Learning', 'Artificial Intelligence',
    'Data Science', 'Cybersecurity', 'Linux'
]

skill_demand = {}

for skill in cse_skills:
    skill_demand[skill] = skills.str.contains(
        skill, case=False, na=False
    ).sum()

cse_skill_demand = (
    pd.Series(skill_demand)
    .sort_values(ascending=False)
)

cse_skill_demand
```

```python
cse_skill_demand.plot(
    kind='bar',
    figsize=(12, 6),
    title='CSE/IT Skills Demand'
)
```

```python
salary_experience = df[
    ['minimumExperience', 'maximumExperience',
     'minimumSalary', 'maximumSalary']
].dropna()

salary_experience['averageSalary'] = (
    salary_experience['minimumSalary'] +
    salary_experience['maximumSalary']
) / 2

salary_experience.groupby('minimumExperience')['averageSalary'].mean()
```

```python
salary_by_exp = (
    salary_experience
    .groupby('minimumExperience')['averageSalary']
    .mean()
)

salary_by_exp.plot(
    kind='line',
    figsize=(12, 6),
    title='Average Salary by Experience'
)
```
